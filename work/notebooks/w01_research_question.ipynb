**Provisional lane: Lane 2 — Refresh / Content Opportunity Scoring.**

```python
# Load the starter data and show basic supporting numbers.
# This cell uses only observable starter-data fields.

import pandas as pd

df = pd.read_csv("/content/content_refresh_anonymized.csv")

print(f"Rows: {len(df):,}")
print(f"Unique content pages: {df['content_id'].nunique():,}")
print(f"Unique clients: {df['client_id'].nunique():,}")

declining_pct = (df["trend_direction"].eq("down").mean() * 100)
median_impressions = df["impressions_90d"].median()
visible_500_pct = (df["impressions_90d"] >= 500).mean() * 100

print(f"Pages with downward current-window trend: {declining_pct:.1f}%")
print(f"Median 90-day impressions per page: {median_impressions:,.0f}")
print(f"Pages with at least 500 90-day impressions: {visible_500_pct:.1f}%")
```

## 2. The question: decision, action, cost of a wrong call

**Research question:** When review capacity is limited, which content pages should be reviewed first based on observable evidence?

- **Decision:** Decide which pages receive review attention first.
- **Who acts:** A content/SEO reviewer or editor uses the ranked queue to decide where to spend limited review time.
- **Action:** The reviewer may choose to refresh, expand, protect, prune, or monitor a page after inspecting the evidence.
- **Unit of analysis:** One content page (`content_id`).
- **Output:** A ranked priority score/queue, ideally with reason codes that explain why a page is high priority.
- **Cost of a wrong call:** A false positive can waste scarce reviewer time on a page that was not a good candidate. A false negative can cause a useful review opportunity to be missed. The relative cost of these errors should influence the final ranking threshold and validation metric.

This is a ranking/decision-support problem rather than simply a request to predict whether a page is "bad." The goal is to improve the order in which humans review pages, not to automate the final content decision.
